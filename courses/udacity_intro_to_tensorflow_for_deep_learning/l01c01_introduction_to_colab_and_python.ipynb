##Dividing data in to test and train sets

```
print("dropping price col from X data set")
X=df.drop('price',axis=1)
print("successfully dropped price col ")
print("creating y data set")
y=df['price']
print("successfully created y data set")

```

```
from sklearn.model_selection import train_test_split
print("splitting data into train and test")
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=2)
print("successfully splitted data into train and test")
print("displaying X_train")
X_train.head()

```

##model building

```
from sklearn.linear_model import LinearRegression
import pandas as pd

lr_model = LinearRegression()

# Identify categorical columns (columns with 'object' dtype)
categorical_cols = X_train.select_dtypes(include='object').columns

# Apply one-hot encoding to convert categorical features into numerical ones.
# `drop_first=True` avoids multicollinearity by dropping the first category for each feature.
X_train = pd.get_dummies(X_train, columns=categorical_cols, drop_first=True)
X_test = pd.get_dummies(X_test, columns=categorical_cols, drop_first=True)

# Ensure that X_test has the same columns as X_train after one-hot encoding.
# This is crucial if some categories are present in the training set but not in the test set, or vice versa.
# `fill_value=0` will add missing columns and fill them with zeros.
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

print("Training model...")
lr_model.fit(X_train, y_train)
print("Model training successful!")

print("Training data accuracy:", lr_model.score(X_train, y_train))
print("Testing data accuracy:", lr_model.score(X_test, y_test))
```

##prediction

```
print("Generating predictions on X_test...")
y_pred = lr_model.predict(X_test)

print("First 10 predictions:")
print(y_pred[:10])

print("\nActual values for the first 10 test samples:")
print(y_test.head(10))
```
